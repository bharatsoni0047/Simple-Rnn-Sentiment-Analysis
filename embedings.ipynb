```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = ["bharat is a data scientist",
        " i love AI/Ml",
        "python is good langauge",
        "i am using tensorFlow",
        "my favorite movie is iron man 3"]
```

```python
# define vocabulary size
voc_size = 10000
```

```python
one_hot_repr = [one_hot(words,voc_size) for words in sent]
one_hot_repr
```

this is showing index of the words in the vocab eg. > if "bhaart" is matching atindex of 9546 index others are 0

```python
# word Embeding Representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
#from tensorflow.keras.utils import pad_sequences
import numpy as np
```

```python
#setting max sent length
sent_len = 8
embeded_docs = pad_sequences(one_hot_repr, padding= 'pre', maxlen= sent_len)
print(embeded_docs)
```

```python
# Feature representation
dim = 10 # there are 300 dimensions in the trained model

model = Sequential()
model.add(Embedding(voc_size,dim,input_length = sent_len))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embeded_docs)
```
